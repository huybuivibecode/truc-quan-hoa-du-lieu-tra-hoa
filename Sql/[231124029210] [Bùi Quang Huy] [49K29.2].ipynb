### 1. Năm qua, mặt hàng nào bán chạy nhất? mặt hàng nào bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã mặt hàng, Tên mặt hàng, Số lượng bán, Doanh số bán.</i>

```python
%%sql
SELECT
    p.product_code as 'Mã mặt hàng',
    p.product_name as 'Tên mặt hàng',
    sum(bl.quantity) as 'Số lượng bán',
    sum(bl.quantity*p.price) as 'Doanh số bán'
FROM product p
JOIN bill_line bl on bl.product_id = p.product_id
GROUP BY p.product_code, p.product_name
ORDER BY sum(bl.quantity*p.price) DESC;
```

### 2. Năm qua, nhóm hàng nào bán chạy nhất? nhóm hàng nào bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã nhóm hàng, Tên nhóm hàng, Số lượng bán, Doanh số bán.</i>

```python
%%sql

SELECT
    c.category_code as 'Mã nhóm hàng',
    c.category_name as 'Tên nhóm hàng',
    sum(bl.quantity) as 'Số lượng bán',
    sum(bl.quantity*p.price) as 'Doanh số bán'
FROM category c
JOIN product p on p.category_id = c.category_id
JOIN bill_line bl on bl.product_id = p.product_id
GROUP BY c.category_code, c.category_name
ORDER BY sum(bl.quantity*p.price) DESC;
```

### 3. Năm qua, tháng nào bán chạy nhất? tháng nào bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Tháng, Số lượng bán, Doanh số bán.</i>

```python
%%sql

SELECT
    strftime('%Y-%m', b.time_created) as "Tháng",
    sum(bl.quantity) as "Số lượng bán",
    sum(bl.quantity * p.price) as "Doanh số bán"
FROM bill b
JOIN bill_line bl ON b.bill_id = bl.bill_id
JOIN product p ON bl.product_id = p.product_id
WHERE strftime('%Y', b.time_created) = '2022'
GROUP BY "Tháng"
ORDER BY "Doanh số bán" DESC;
```

### 4. Năm qua, ngày nào trong tuần thường bán chạy nhất? ngày nào trong tuần thường bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Ngày trong tuần, Số lượng bán Trung bình, Doanh số bán Trung bình.</i>

```python
%%sql
SELECT
    CASE strftime('%w', b.time_created)
    WHEN '0' THEN 'Chủ nhật'
    WHEN '1' THEN 'Thứ hai'
    WHEN '2' THEN 'Thứ ba'
    WHEN '3' THEN 'Thứ tư'
    WHEN '4' THEN 'Thứ năm'
    WHEN '5' THEN 'Thứ sáu'
    WHEN '6' THEN 'Thứ bảy'
    END
    AS 'Ngày trong tuần',
    SUM(bl.quantity) / 52 AS 'Số lượng bán',
    SUM(bl.quantity * p.price) / 52 AS 'Doanh số bán'
FROM bill b
JOIN bill_line bl ON b.bill_id = bl.bill_id
JOIN product p ON bl.product_id = p.product_id
WHERE strftime('%Y', b.time_created) = '2022'
GROUP BY strftime('%w', b.time_created)
ORDER BY SUM(bl.quantity) DESC;
```

### 5. Năm qua, ngày nào trong tháng thường bán chạy nhất? ngày nào trong tháng thường bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Ngày trong tháng, Số lượng bán trung bình, Doanh số bán trung bình.</i>

```python
%%sql
SELECT
    strftime('%d', b.time_created) AS 'Ngày trong tháng',
    SUM(bl.quantity) / 12 AS 'Số lượng bán',
    SUM(bl.quantity * p.price) / 12 AS 'Doanh số bán'
FROM bill b
JOIN bill_line bl ON b.bill_id = bl.bill_id
JOIN product p ON bl.product_id = p.product_id
WHERE strftime('%Y', b.time_created) = '2022'
GROUP BY strftime('%d', b.time_created)
ORDER BY SUM(bl.quantity * p.price) / 12 DESC;
```

### 6. Năm qua, khung giờ nào thường bán chạy nhất? khung giờ nào thường bán kém nhất?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Khung giờ, Số lượng bán trung bình, Doanh số bán trung bình.</i>

```python
%%sql
SELECT
    strftime('%H:00 - %H:59', b.time_created) AS 'Khung giờ',
    SUM(bl.quantity) / 365 AS "Số lượng bán trung bình",
    SUM(bl.quantity * p.price) / 365 AS "Doanh số bán trung bình"
FROM bill b
JOIN bill_line bl ON b.bill_id = bl.bill_id
JOIN product p ON bl.product_id = p.product_id
GROUP BY strftime('%H', b.time_created)
ORDER BY SUM(bl.quantity * p.price) / 365 DESC;
```

### 7. Năm qua, nhóm hàng nào thường dễ bán nhất (có xác suất bán trên mỗi đơn hàng cao nhất)? nhóm hàng nào khó bán nhất (có xác suất bán trên mỗi đơn hàng thấp nhất)?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã nhóm hàng, Tên nhóm hàng, Xác suất bán trên đơn hàng.</i>

```python
%%sql
WITH total_bills AS (
    SELECT COUNT(DISTINCT b.bill_id) AS total_count
    FROM bill b
)

SELECT
    c.category_code AS 'Mã nhóm hàng',
    c.category_name AS 'Tên nhóm hàng',
    ROUND(COUNT(DISTINCT bl.bill_id) * 1.0 / (SELECT total_count FROM total_bills), 2) || "%" as 'Xác suất bán trên đơn hàng'
FROM category c
JOIN bill_line bl ON bl.product_id = p.product_id
JOIN product p ON c.category_id = p.category_id
JOIN bill b ON b.bill_id = bl.bill_id
GROUP BY c.category_id, c.category_name
ORDER BY count(DISTINCT bl.bill_id) * 1.0 / (SELECT total_count FROM total_bills) DESC;

```

### 8. Năm qua, xác xuất mua hàng (trên mỗi đơn hàng) của mỗi nhóm hàng biến thiên như thế nào theo từng tháng trong năm?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã nhóm hàng, Tên nhóm hàng, Xác suất bán trên đơn hàng của tháng cao nhất, thấp nhất trong năm, xu hướng tăng giảm của xác suất bán.</i>

```python
%%sql

WITH monthly_category_bills AS (
    SELECT
        strftime('%Y-%m', b.time_created) AS "Tháng",
        c.category_code AS "Mã nhóm hàng",
        c.category_name AS "Tên nhóm hàng",
        COUNT(DISTINCT b.bill_id) AS category_bill_count
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    JOIN category c ON p.category_id = c.category_id
    GROUP BY 1, 2, 3
),
monthly_total_bills AS (
    SELECT
        strftime('%Y-%m', b.time_created) AS "Tháng",
        COUNT(DISTINCT b.bill_id) AS total_bill_count
    FROM bill b
    GROUP BY 1
)
SELECT
    SUBSTR(mcb."Tháng", 6, 2) AS "Tháng",
    MAX(CASE WHEN mcb."Mã nhóm hàng" = 'BOT' THEN ROUND(CAST(mcb.category_bill_count AS REAL) * 100 / mtb.total_bill_count, 1) || '%' ELSE NULL END) AS "[BOT] Bột",
    MAX(CASE WHEN mcb."Mã nhóm hàng" = 'SET' THEN ROUND(CAST(mcb.category_bill_count AS REAL) * 100 / mtb.total_bill_count, 1) || '%' ELSE NULL END) AS "[SET] Set trà",
    MAX(CASE WHEN mcb."Mã nhóm hàng" = 'THO' THEN ROUND(CAST(mcb.category_bill_count AS REAL) * 100 / mtb.total_bill_count, 1) || '%' ELSE NULL END) AS "[THO] Trà hoa",
    MAX(CASE WHEN mcb."Mã nhóm hàng" = 'TMX' THEN ROUND(CAST(mcb.category_bill_count AS REAL) * 100 / mtb.total_bill_count, 1) || '%' ELSE NULL END) AS "[TMX] Trà mix",
    MAX(CASE WHEN mcb."Mã nhóm hàng" = 'TTC' THEN ROUND(CAST(mcb.category_bill_count AS REAL) * 100 / mtb.total_bill_count, 1) || '%' ELSE NULL END) AS "[TTC] Trà củ, quả sấy"
FROM monthly_category_bills mcb
JOIN monthly_total_bills mtb ON mcb."Tháng" = mtb."Tháng"
GROUP BY mcb."Tháng"
ORDER BY mcb."Tháng";
```

### 9. Năm qua, đối với từng nhóm hàng thì mặt hàng nào thường dễ bán nhất (có xác suất bán trên mỗi đơn hàng cao nhất)? mặt hàng nào khó bán nhất (có xác suất bán trên mỗi đơn hàng thấp nhất)?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã nhóm hàng, Tên nhóm hàng, Mã mặt hàng, Tên mặt hàng, Xác suất bán trên đơn hàng.</i>

```python
%%sql
WITH category_bills AS (
    SELECT
        c.category_id,
        COUNT(DISTINCT b.bill_id) AS total_category_bills
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    JOIN category c ON p.category_id = c.category_id
    GROUP BY c.category_id
),
product_category_bills AS (
    SELECT
        c.category_id,
        c.category_code,
        c.category_name,
        p.product_id,
        p.product_code,
        p.product_name,
        COUNT(DISTINCT b.bill_id) AS product_bill_count
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    JOIN category c ON p.category_id = c.category_id
    GROUP BY c.category_id, c.category_code, c.category_name, p.product_id, p.product_code, p.product_name
)
SELECT
    pcb.category_code AS "Mã nhóm hàng",
    pcb.category_name AS "Tên nhóm hàng",
    pcb.product_code AS "Mã mặt hàng",
    pcb.product_name AS "Tên mặt hàng",
    ROUND(CAST(pcb.product_bill_count AS REAL) * 100 / cb.total_category_bills, 2) || "%" AS "Xác suất bán trên đơn hàng"
FROM product_category_bills pcb
JOIN category_bills cb ON pcb.category_id = cb.category_id
ORDER BY pcb.category_name, "Xác suất bán trên đơn hàng" DESC;
```

### 10. Năm qua, xác xuất mua hàng (trên mỗi đơn hàng) của mỗi mặt hàng trong mỗi nhóm hàng biến thiên như thế nào theo từng tháng trong năm?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Mã nhóm hàng, Tên nhóm hàng, Mã mặt hàng, Tên mặt hàng, Xác suất bán trên đơn hàng của tháng cao nhất, thấp nhất trong năm, xu hướng tăng giảm của xác suất bán.</i>

```python
%%sql

WITH monthly_category_bills AS (
    SELECT
        strftime('%Y-%m', b.time_created) AS "Tháng",
        c.category_id,
        COUNT(DISTINCT b.bill_id) AS total_category_bills_month
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    JOIN category c ON p.category_id = c.category_id
    GROUP BY 1, 2
),
monthly_product_bills AS (
    SELECT
        strftime('%Y-%m', b.time_created) AS "Tháng",
        c.category_code AS "Mã nhóm hàng",
        c.category_name AS "Tên nhóm hàng",
        p.product_code AS "Mã mặt hàng",
        p.product_name AS "Tên mặt hàng",
        c.category_id,
        COUNT(DISTINCT b.bill_id) AS product_bill_count_month
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    JOIN category c ON p.category_id = c.category_id
    GROUP BY 1, 2, 3, 4, 5, 6
)
SELECT
    SUBSTR(mpb."Tháng", 6, 2) AS "Tháng",
    mpb."Mã nhóm hàng",
    mpb."Tên nhóm hàng",
    mpb."Mã mặt hàng",
    mpb."Tên mặt hàng",
    mpb.product_bill_count_month AS "SL Đơn Bán",
    mcb.total_category_bills_month AS "Số Đơn Bán của Nhóm hàng trong tháng",
    ROUND(CAST(mpb.product_bill_count_month AS REAL) * 100 / mcb.total_category_bills_month, 1) || "%" AS "Xác suất Bán"
FROM monthly_product_bills mpb
JOIN monthly_category_bills mcb ON mpb."Tháng" = mcb."Tháng" AND mpb.category_id = mcb.category_id
ORDER BY mpb."Tháng", mpb."Mã nhóm hàng", "Xác suất Bán" DESC;
```

### 11. Năm qua, phân phối mức độ mua lặp lại của khách hàng như thế nào?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Số lần mua lặp lại, mức độ phân bổ của các mức mua lặp lại của khách hàng.</i>

```python
%%sql

WITH customer_bill_count AS (
    SELECT
        customer_id,
        COUNT(bill_id) AS num_bills
    FROM bill
    GROUP BY customer_id
)
SELECT
    num_bills AS 'Số lần mua',
    COUNT(customer_id) AS 'Số lượng khách hàng'
FROM customer_bill_count
GROUP BY num_bills
ORDER BY num_bills;
```

### 12. Năm qua, phân phối số tiền của khách hàng đã chi trả cho doanh nghiệp như thế nào?. Trình bày dữ liệu dạng bảng và đưa ra nhận định của bản thân lý giải nguyên do của kết quả.
<i>*Note: Trả lời thành văn bản trong đó có thông tin Số tiền khách hàng đã chi trả, mức độ phân bổ của các mức mua lặp lại của khách hàng.</i>

```python
%%sql

WITH customer_total_spent AS (
    SELECT
        b.customer_id,
        SUM(bl.quantity * p.price) AS total_spent
    FROM bill b
    JOIN bill_line bl ON b.bill_id = bl.bill_id
    JOIN product p ON bl.product_id = p.product_id
    GROUP BY b.customer_id
),
binning AS (
    SELECT
        customer_id,
        total_spent,
        CAST(FLOOR(total_spent / 50000) * 50000 AS INTEGER) AS bin_start,
        CAST((FLOOR(total_spent / 50000) + 1) * 50000 AS INTEGER) AS bin_end
    FROM customer_total_spent
)
SELECT
    'Từ ' || bin_start || ' đến ' || bin_end || ' VNĐ' AS "Khoảng chi tiêu",
    COUNT(customer_id) AS "Số lượng khách hàng"
FROM binning
GROUP BY bin_start, bin_end
ORDER BY bin_start;
```
